```python
# %% Downsample
import os
import h5py

FACTOR = 20
in_file = "CSTR_SimulationData.h5"
out_file = f"CSTR_SimulationData_ds{FACTOR}.h5"

if os.path.exists(out_file):
    os.remove(out_file)

with h5py.File(in_file, "r") as f_in, h5py.File(out_file, "w") as f_out:
    for name in f_in.keys():
        data = f_in[name][::FACTOR]
        f_out.create_dataset(name, data=data, chunks=True, compression="gzip")
        print(f"{name}: {f_in[name].shape} -> {data.shape}")
```
